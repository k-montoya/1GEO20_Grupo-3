## **Descripción del proyecto**

> El proyecto analiza la variabilidad interanual y espacial de la temperatura del aire a 2 metros en el departamento de Piura durante el periodo 2000–2025 (312 meses) y evalúa su relación temporal y estadística con El Niño y La Niña.  Los datos térmicos provienen del reanálisis terrestre ERA5-Land (ECMWF/ERA5_LAND/MONTHLY_AGGR), que cubre exclusivamente superficie terrestre, y se promedian sobre un rectángulo delimitador (Bounding Box) sobre el dominio geográfico de Piura. La condición oceánica se sigue con dos índices independientes, el Índice Oceánico El Niño (ONI) del Pacífico central y el Índice Costero El Niño (ICEN) del mar frente a Perú y Ecuador.


**Objetivos:**

Objetivo general: Analizar la variabilidad interanual y la distribución espacial de la temperatura del aire a 2 metros en Piura entre 2000 y 2025, y evaluar su relación con las fases de El Niño y La Niña.

Objetivos específicos:


1.   Caracterizar la climatología mensual, la variabilidad interanual y la distribución espacial de la temperatura del aire a 2 metros en el dominio de Piura a partir de datos de ERA5-Land.
2. Evaluar la relación temporal entre las anomalías de temperatura en Piura y los índices ONI e ICEN mediante correlación (lineal y no paramétrica), desfases (lags) y distribución por fases.



**Hipótesis exploratoria:**

Se espera que las anomalías de temperatura del aire en Piura presenten valores positivos (por encima de su climatología mensual) durante los meses clasificados en fase El Niño, y valores negativos durante las fases La Niña, mostrando un acoplamiento más estrecho con el índice costero (ICEN) debido a la proximidad geográfica del departamento a la región Niño 1+2. Esta expectativa se sustenta en registros previos como El Niño costero de 2017, cuando las temperaturas del aire en la costa norte peruana superaron entre 1,0 y 2,3 °C sus valores normales (Takahashi et al., 2018).

Se considerará que la hipótesis se cumple si (i) la mediana de la anomalía es positiva en fase El Niño y negativa en fase La Niña, y (ii) la correlación con el ICEN es mayor, en valor absoluto, que la correlación con el ONI.

**Variables a analizar:**

1.   Temperatura del aire a 2 mt:

      *   Promedio mensual de datos horarios sobre superficie terrestre con resolución espacial nativa de ~9–11 km, extraído mediante Bounding Box ([-81.5, -6.5, -78.5, -3.5]) y transformado de Kelvin a grados Celsius (°C).




2.   Condición del El Niño-Oscilación del Sur (usando dos índices):

      *   Índice Oceánico El Niño (ONI): promedio de tres meses de la anomalía de temperatura superficial en el Pacífico Central (región Niño 3.4). Un valor de +0,5 °C o más indica El Niño y uno de −0,5 °C o menos indica La Niña (CPC, s.f.).
      

      *   Índice Costero El Niño (ICEN): la misma medida para el mar frente a Perú y Ecuador (región Niño 1+2). Las condiciones cálidas empiezan cuando se supera +0,5 °C y las frías cuando baja de −0,7 °C (ENFEN, 2024).

**Área de estudio:**
> El área de estudio es el departamento de Piura, en el noroeste del Perú. Tiene 35 892 km², limita con Tumbes, Ecuador, Cajamarca, Lambayeque y el océano Pacífico. Su relieve va del desierto de Sechura, con depresiones bajo el nivel del mar, a cerca de 3900 m en la cordillera Occidental (Rodríguez et al., 2011).

> La temperatura se promedia sobre un rectángulo y no sobre el límite del departamento. El rectángulo va de 3,5° a 6,5° S y de 78,5° a 81,5° O y cubre unos 110 000 km², unas tres veces la superficie de Piura. Además de todo el departamento, abarca parte de Tumbes, Lambayeque, Cajamarca y Amazonas, junto con zonas de las provincias ecuatorianas de El Oro, Loja y Zamora Chinchipe.

> Piura se eligió por su exposición a El Niño. Su llanura costera es árida en años normales, pero durante El Niño recibe lluvias intensas asociadas al calentamiento del mar cercano (Takahashi, 2004). Ese mar forma parte de la región Niño 1+2, la que mide el ICEN, por lo que Piura permite evaluar si su calentamiento también se refleja en la temperatura del aire.

**Periodo temporal:**

El periodo va de enero de 2000 a diciembre de 2025 (312 meses) y se cierra en 2025 porque es el último año completo en todas las series. En estos 26 años, el ONI registra ocho episodios El Niño y diez La Niña, contando La Niña que ya estaba en curso a inicios de 2000 (CPC, s. f.). Frente a la costa, la serie oficial del ICEN muestra diez eventos El Niño costero y once La Niña costera, con el criterio de al menos tres meses seguidos en condición cálida o fría (ENFEN, 2024).

El evento cálido más intenso en el Pacífico central fue El Niño de 2015-2016, con un ONI máximo de +2,6 °C y magnitud fuerte también frente a la costa. En la costa, el mayor fue El Niño costero de 2023-2024, con un ICEN de +2,92 °C en julio de 2023. Entre las fases frías destacan La Niña de 2007-2008 (ONI mínimo de -1,8 °C, con La Niña costera fuerte entre abril y diciembre de 2007) y la secuencia casi continua de La Niña entre 2020 y 2023, visible en ambos índices.


La Niña costera fuerte de 2013 (ICEN de −1,66 °C en junio) y El Niño costero de 2017 (+1,31 °C en marzo) ocurrieron con el ONI en rango neutro. Sin embargo, a inicios del 2000, el ONI marcaba La Niña (-1,5 °C) sin que hubiera un evento frente a la costa.

### Variable 1 : Temperatura Promedio Mensual del Aire a 2mt - ERA5 Land

```python
# Archivo vectorial del departamento de Piura
bbox_piura = [-81.5, -6.5, -78.5, -3.5]
roi_bbox_era5 = ee.Geometry.Rectangle(bbox_piura)

# Cargamos la colección mensual de ERA5-Land, filtramos por nuestra área y periodo, y seleccionamos la banda t2m
era5_mensual = (ee.ImageCollection("ECMWF/ERA5_LAND/MONTHLY_AGGR")
                .filterBounds(roi_bbox_era5)
                .filterDate("2000-01-01", "2025-12-31")
                .select("temperature_2m"))

# Función para extraer el promedio espacial de la temperatura por cada imagen mensual
def reducir_region_piura(img):
    mean_val = img.reduceRegion(
        reducer=ee.Reducer.mean(), # Media espacial de los píxeles dentro del departamento
        geometry=roi_bbox_era5,
        scale=11132, # Resolución nativa aproximada de ERA5-Land (~9-11 km)
        maxPixels=1e9 # Aumentamos el límite de píxeles procesables
    ).get("temperature_2m")

    # Retornamos un Feature (entidad) que almacena la fecha y el valor térmico en Kelvin
    return ee.Feature(None, {
        'date': img.date().format('YYYY-MM-dd'),
        't2m_K': mean_val
    })

# Mapeamos la función sobre toda la colección y extraemos los resultados hacia un DataFrame
# .map() aplica la función a cada imagen, .filter() quita resultados nulos, y .getInfo() extrae los datos de la nube a Python
features_era5 = era5_mensual.map(reducir_region_piura).filter(ee.Filter.notNull(['t2m_K'])).getInfo()['features']
df_era5 = pd.DataFrame([f['properties'] for f in features_era5]) # Conversión a DataFrame de Pandas

# Conversión de unidades (Kelvin a Celsius) y extracción de variables temporales para el gráfico
df_era5['fecha'] = pd.to_datetime(df_era5['date'])
df_era5['valor_C_era5'] = df_era5['t2m_K'] - 273.15
df_era5['año'] = df_era5['fecha'].dt.year
df_era5['mes'] = df_era5['fecha'].dt.month
print(f"Meses extraídos de ERA5-Land: {len(df_era5)} (esperado: 312)")

# Reestructuramos los datos con pivot: meses en el eje X, y cada año será una columna independiente
df_pivot_era5 = df_era5.pivot(index='mes', columns='año', values='valor_C_era5')

# Exportamos el DataFrame en crudo a Google Drive
# Guardamos el archivo CSV específico de ERA5, omitiendo el índice automático de las filas (index=False)
df_era5.to_csv('/content/drive/My Drive/ERA5_T2M_Piura.csv', index=False)
print("Archivo crudo ERA5 guardado exitosamente en Google Drive.")
```

```python
# Gráfico de líneas: ciclo anual de cada año (el color indica el año)
etiquetas_mes = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
anios = list(df_pivot_era5.columns)
cmap_anios = plt.cm.viridis

fig_era5, ax_era5 = plt.subplots(figsize=(10, 5))
for i, col in enumerate(anios):
    ax_era5.plot(df_pivot_era5.index, df_pivot_era5[col], color=cmap_anios(i / (len(anios) - 1)), linewidth=1.2, alpha=0.85)

sm = plt.cm.ScalarMappable(cmap=cmap_anios, norm=plt.Normalize(anios[0], anios[-1]))
fig_era5.colorbar(sm, ax=ax_era5, label='Año')

ax_era5.set_xticks(range(1, 13))
ax_era5.set_xticklabels(etiquetas_mes)
ax_era5.set_title('Temperatura Promedio Mensual a 2m (ERA5-Land) - Departamento de Piura (2000-2025)', fontsize=12)
ax_era5.set_xlabel('Meses', fontsize=11)
ax_era5.set_ylabel('Temperatura (°C)', fontsize=11)
ax_era5.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Generación del mapa

# Promediamos toda la serie histórica y la cortamos a la zona de estudio
img_promedio_era5 = era5_mensual.mean().subtract(273.15).clip(roi_bbox_era5)

# Descargamos el GeoTIFF usando el límite del polígono como región
url_era5 = img_promedio_era5.getDownloadURL({
    "region": roi_bbox_era5,
    "scale": 11132,
    "crs": "EPSG:4326",
    "format": "GEO_TIFF"
})
r_era5 = requests.get(url_era5, timeout=300)
with open("t2m_piura_era5.tif", "wb") as f_era5:
    f_era5.write(r_era5.content)


# Leemos el GeoTIFF con Rioxarray y lo preparamos para Cartopy
tif_era5 = rioxarray.open_rasterio("t2m_piura_era5.tif").squeeze("band", drop=True)
fig_mapa_era5, ax_mapa_era5 = plt.subplots(figsize=(8, 8), subplot_kw={'projection': ccrs.PlateCarree()})


# Graficamos la matriz de datos con contornos de colores

tif_era5.plot.contourf(
    ax=ax_mapa_era5,
    transform=ccrs.PlateCarree(),
    levels=np.arange(10, 30, 1),
    cmap='coolwarm',
    cbar_kwargs={
        'orientation': 'horizontal',
        'shrink': 0.7,
        'pad': 0.08,
        'label': 'T2m Promedio (°C)'
    }
)

ax_mapa_era5.set_extent([-82, -78.5, -6.5, -3.5], crs=ccrs.PlateCarree())

# Agregamos los elementos cartográficos específicos

ax_mapa_era5.add_feature(cfeature.COASTLINE) # Línea de costa
ax_mapa_era5.add_feature(cfeature.BORDERS, linestyle='-') # Frontera con Ecuador
ax_mapa_era5.add_feature(cfeature.STATES, edgecolor="gray", linewidth=0.8) # Límites departamentales vecinos
ax_mapa_era5.add_feature(cfeature.OCEAN, facecolor='lightcyan', zorder=10) # Océano

ax_mapa_era5.gridlines(draw_labels=True, color='gray', alpha=0.5, linestyle='--')
ax_mapa_era5.set_title('Distribución Espacial de T2m Promedio - Departamento de Piura (2000-2025)')
plt.show()
```

### Resultado 1 : Climatología, Anomalías Mensuales y Variabilidad Interanual de la Temperatura del Aire a 2mt en Piura (2000 - 2025)

```python
# Climatología mensual: promedio 2000-2025 de cada mes y su variabilidad entre años
clim_tabla = df_era5.groupby('mes')['valor_C_era5'].agg(['mean', 'std'])

fig_clim, ax_clim = plt.subplots(figsize=(9, 4.5))
ax_clim.plot(clim_tabla.index, clim_tabla['mean'], color='black', marker='o', label='Climatología 2000–2025')
ax_clim.fill_between(clim_tabla.index, clim_tabla['mean'] - clim_tabla['std'], clim_tabla['mean'] + clim_tabla['std'],
                     color='gray', alpha=0.3, label='±1 desviación estándar entre años')
ax_clim.set_xticks(range(1, 13))
ax_clim.set_xticklabels(etiquetas_mes)
ax_clim.set_title('Climatología Mensual de T2m (ERA5-Land) - Piura (2000-2025)', fontsize=12)
ax_clim.set_xlabel('Meses', fontsize=11)
ax_clim.set_ylabel('Temperatura (°C)', fontsize=11)
ax_clim.grid(True, linestyle=':', alpha=0.6)
ax_clim.legend()
plt.tight_layout()
plt.show()

print(clim_tabla.round(2).rename(columns={'mean': 'Media (°C)', 'std': 'Desv. estándar (°C)'}))
```

```python
# Calculamos la climatología mensual
# Agrupamos por la columna 'mes' y promediamos 'valor_C_era5' a lo largo de todo el periodo 2000-2025
climatologia_mensual = df_era5.groupby('mes')['valor_C_era5'].transform('mean')

# Calculamos la anomalía térmica mensual (°C)
# Fórmula: Anomalía = Temperatura observada del mes - Climatología de ese mismo mes
df_era5['anomalia_T2m'] = df_era5['valor_C_era5'] - climatologia_mensual

# Ordenamos cronológicamente para asegurar continuidad en la serie temporal
df_era5 = df_era5.sort_values('fecha').reset_index(drop=True)

# Visualización de la serie de tiempo de anomalías térmicas (2000-2025)
fig_anom, ax_anom = plt.subplots(figsize=(12, 4.5))

# Coloreamos las barras o áreas según si la anomalía es positiva (cálida) o negativa (fría)
ax_anom.fill_between(df_era5['fecha'], df_era5['anomalia_T2m'], 0,
                     where=(df_era5['anomalia_T2m'] >= 0), color='crimson', alpha=0.6, interpolate=True, label='Anomalía positiva (+°C)')
ax_anom.fill_between(df_era5['fecha'], df_era5['anomalia_T2m'], 0,
                     where=(df_era5['anomalia_T2m'] < 0), color='royalblue', alpha=0.6, interpolate=True, label='Anomalía negativa (-°C)')

# Trazamos la línea negra de la serie y la línea base en 0 °C
ax_anom.plot(df_era5['fecha'], df_era5['anomalia_T2m'], color='black', linewidth=0.8)
ax_anom.axhline(0, color='black', linestyle='--', linewidth=1)

# Formato del gráfico
ax_anom.set_title('Anomalías Mensuales de Temperatura a 2m (ERA5-Land) - Departamento de Piura (2000-2025)', fontsize=12)
ax_anom.set_xlabel('Año', fontsize=11)
ax_anom.set_ylabel('Anomalía Térmica (°C)', fontsize=11)
ax_anom.grid(True, linestyle=':', alpha=0.6)
ax_anom.legend(loc='upper left')

plt.tight_layout()
plt.show()

# Exportamos la tabla actualizada con climatología y anomalías a Google Drive
df_era5.to_csv('/content/drive/My Drive/ERA5_T2M_Anomalias_Piura_2000_2025.csv', index=False)
print("Serie de anomalías térmicas calculada y guardada exitosamente.")
```

```python
# Variabilidad interanual: anomalía media de cada año, su dispersión y su tendencia
anomalia_anual = df_era5.groupby('año')['anomalia_T2m'].mean()
pend_anual, inter_anual = np.polyfit(anomalia_anual.index, anomalia_anual.values, 1)  # °C por año

fig_anual, ax_anual = plt.subplots(figsize=(12, 4.5))
ax_anual.bar(anomalia_anual.index, anomalia_anual.values,
             color=np.where(anomalia_anual.values >= 0, 'crimson', 'royalblue'), alpha=0.7)
ax_anual.plot(anomalia_anual.index, pend_anual * anomalia_anual.index + inter_anual, color='black', linestyle='--',
              label=f'Tendencia lineal ({pend_anual * 10:+.2f} °C por década)')
ax_anual.axhline(0, color='black', linewidth=0.8)
ax_anual.set_title('Anomalía Media Anual de T2m (ERA5-Land) - Piura (2000-2025)', fontsize=12)
ax_anual.set_xlabel('Año', fontsize=11)
ax_anual.set_ylabel('Anomalía media anual (°C)', fontsize=11)
ax_anual.grid(True, linestyle=':', alpha=0.6)
ax_anual.legend(loc='upper left')
plt.tight_layout()
plt.show()

print(f"Desviación estándar de las anomalías mensuales: {df_era5['anomalia_T2m'].std():.2f} °C")
print(f"Desviación estándar de las medias anuales: {anomalia_anual.std():.2f} °C")
print(f"Año más cálido: {anomalia_anual.idxmax()} ({anomalia_anual.max():+.2f} °C)")
print(f"Año más frío: {anomalia_anual.idxmin()} ({anomalia_anual.min():+.2f} °C)")
print(f"Tendencia lineal: {pend_anual * 10:+.2f} °C por década")
```

### Resultado 4 : Correlación Cruzada con Desfases Temporales entre las Anomalías de T° y los Índices ENSO

```python
# Lags (desfaso) de 6 meses antes y 6 meses después
lags = range(-6, 7) # De -6 meses (índice adelantado) a +6 meses (índice rezagado)
corrs_icen = []
corrs_oni = []

for lag in lags:
    # Desplazamos los índices respecto a la temperatura de Piura
    # lag < 0: el océano se calienta ANTES que el aire en Piura
    df_temp = pd.DataFrame({
        'T2m': df_final['anomalia_T2m'],
        'ICEN_shift': df_final['ICEN'].shift(-lag),
        'ONI_shift': df_final['ONI'].shift(-lag)
    }).dropna()

    r_i, _ = stats.pearsonr(df_temp['T2m'], df_temp['ICEN_shift'])
    r_o, _ = stats.pearsonr(df_temp['T2m'], df_temp['ONI_shift'])
    corrs_icen.append(r_i)
    corrs_oni.append(r_o)

fig_lag, ax_lag = plt.subplots(figsize=(10, 4.5))
ax_lag.plot(lags, corrs_icen, marker='o', color='crimson', linewidth=2, label='ICEN vs. T2m Piura')
ax_lag.plot(lags, corrs_oni, marker='s', color='royalblue', linewidth=2, label='ONI vs. T2m Piura')
ax_lag.axvline(0, color='black', linestyle='--', linewidth=1, label='Sin desfase (Lag = 0)')

# Marcamos el máximo de cada curva
max_lag_icen = lags[np.argmax(corrs_icen)]
max_lag_oni = lags[np.argmax(corrs_oni)]
ax_lag.annotate(f'Máx ICEN: r={max(corrs_icen):.3f} (Lag {max_lag_icen})',
                xy=(max_lag_icen, max(corrs_icen)), xytext=(max_lag_icen + 0.5, max(corrs_icen) - 0.04),
                arrowprops=dict(arrowstyle='->', color='crimson'), fontsize=9, fontweight='bold', color='darkred')
ax_lag.annotate(f'Máx ONI: r={max(corrs_oni):.3f} (Lag {max_lag_oni})',
                xy=(max_lag_oni, max(corrs_oni)), xytext=(max_lag_oni + 0.5, max(corrs_oni) - 0.08),
                arrowprops=dict(arrowstyle='->', color='royalblue'), fontsize=9, fontweight='bold', color='navy')

ax_lag.set_xticks(lags)
ax_lag.set_title('Correlación Cruzada con Desfase Temporal: Índices ENSO vs. Anomalía T2m en Piura (2000–2025)', fontsize=11, fontweight='bold')
ax_lag.set_xlabel('Desfase en meses (Negativo = Índice oceánico precede a la temperatura en Piura)', fontsize=10)
ax_lag.set_ylabel('Coeficiente de Pearson (r)', fontsize=11)
ax_lag.grid(True, linestyle=':', alpha=0.6)
ax_lag.legend(loc='lower left')
plt.tight_layout()
plt.savefig('grafico3_correlacion_desfases_piura.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Resultado 5 : Distribución de Anomalías de T° por fase ENSO

```python
# Clasificamos cada mes según los umbrales operativos de ICEN y ONI
def clasificar_icen(val):
    if val >= 0.5:
        return 'El Niño\n(ICEN ≥ +0.5)'
    elif val <= -0.7:
        return 'La Niña\n(ICEN ≤ -0.7)'
    else:
        return 'Neutro\n(-0.7 a +0.5)'

def clasificar_oni(val):
    if val >= 0.5:
        return 'El Niño\n(ONI ≥ +0.5)'
    elif val <= -0.5:
        return 'La Niña\n(ONI ≤ -0.5)'
    else:
        return 'Neutro\n(-0.5 a +0.5)'

df_final['Fase_ICEN'] = df_final['ICEN'].apply(clasificar_icen)
df_final['Fase_ONI'] = df_final['ONI'].apply(clasificar_oni)

fig_box, (ax_b1, ax_b2) = plt.subplots(1, 2, figsize=(12, 5), sharey=True)

# Orden lógico de fases: La Niña -> Neutro -> El Niño
orden_icen = ['La Niña\n(ICEN ≤ -0.7)', 'Neutro\n(-0.7 a +0.5)', 'El Niño\n(ICEN ≥ +0.5)']
datos_icen = [df_final[df_final['Fase_ICEN'] == f]['anomalia_T2m'] for f in orden_icen]

bp1 = ax_b1.boxplot(datos_icen, labels=orden_icen, patch_artist=True, widths=0.5)
colores = ['royalblue', 'lightgray', 'crimson']
for patch, color in zip(bp1['boxes'], colores):
    patch.set_facecolor(color)
    patch.set_alpha(0.65)
ax_b1.axhline(0, color='black', linestyle='--', linewidth=0.9)
ax_b1.set_title('Anomalía T2m en Piura según Fase ICEN', fontsize=12, fontweight='bold')
ax_b1.set_ylabel('Anomalía T2m Piura (°C)', fontsize=11)
ax_b1.grid(True, linestyle=':', alpha=0.6)

orden_oni = ['La Niña\n(ONI ≤ -0.5)', 'Neutro\n(-0.5 a +0.5)', 'El Niño\n(ONI ≥ +0.5)']
datos_oni = [df_final[df_final['Fase_ONI'] == f]['anomalia_T2m'] for f in orden_oni]

bp2 = ax_b2.boxplot(datos_oni, labels=orden_oni, patch_artist=True, widths=0.5)
for patch, color in zip(bp2['boxes'], colores):
    patch.set_facecolor(color)
    patch.set_alpha(0.65)
ax_b2.axhline(0, color='black', linestyle='--', linewidth=0.9)
ax_b2.set_title('Anomalía T2m en Piura según Fase ONI', fontsize=12, fontweight='bold')
ax_b2.grid(True, linestyle=':', alpha=0.6)

plt.suptitle('Distribución de Anomalías Térmicas en Piura por Fase ENSO (2000–2025)', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('grafico4_boxplot_fases_enso_piura.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Estadísticos por fase y prueba de diferencias entre fases (Kruskal-Wallis, no paramétrica)
def tabla_fases(col_fase, orden):
    t = (df_final.groupby(col_fase)['anomalia_T2m']
         .agg(n='count', media='mean', mediana='median', desv_est='std')
         .reindex(orden).round(2))
    t.index = [i.replace('\n', ' ') for i in t.index]
    return t

resultados_fases = {}
for nombre, col, orden, datos in [('ICEN', 'Fase_ICEN', orden_icen, datos_icen),
                                  ('ONI', 'Fase_ONI', orden_oni, datos_oni)]:
    print(f"\n=== Anomalía de T2m según fase {nombre} ===")
    print(tabla_fases(col, orden))

    grupos = [d for d in datos if len(d) > 1]
    if len(grupos) >= 2:
        H, p_kw = stats.kruskal(*grupos)
        print(f"Kruskal-Wallis: H = {H:.2f}, p = {p_kw:.2e}")
    else:
        H, p_kw = np.nan, np.nan
        print("Kruskal-Wallis: no hay al menos dos fases con datos suficientes.")

    # Contraste directo entre fases extremas (H1: El Niño > La Niña)
    if len(datos[2]) > 1 and len(datos[0]) > 1:
        U, p_mw = stats.mannwhitneyu(datos[2], datos[0], alternative='greater')
        print(f"Mann-Whitney (El Niño > La Niña): U = {U:.0f}, p = {p_mw:.2e}")
    else:
        p_mw = np.nan
    resultados_fases[nombre] = {'H': H, 'p_kw': p_kw, 'p_mw': p_mw}

# Los meses se clasifican por separado y las series están autocorrelacionadas, por lo que los p-valores son orientativos.
```

### Resultado 6 : Resumen Numérico y Contraste de la Hipótesis

```python
# Anomalía sin la tendencia lineal, para comprobar que la correlación no depende del calentamiento de fondo
t_idx = np.arange(len(df_final))
anom_sin_tend = df_final['anomalia_T2m'] - np.polyval(np.polyfit(t_idx, df_final['anomalia_T2m'], 1), t_idx)

lags_lista = list(lags)
lag_opt_icen = lags_lista[int(np.argmax(corrs_icen))]
lag_opt_oni = lags_lista[int(np.argmax(corrs_oni))]

resumen = pd.DataFrame({
    'Pearson r (lag 0)': [r_icen, r_oni],
    'Kendall tau (lag 0)': [tau_icen, tau_oni],
    'Pearson r sin tendencia': [stats.pearsonr(df_final['ICEN'], anom_sin_tend)[0],
                                stats.pearsonr(df_final['ONI'], anom_sin_tend)[0]],
    'Lag de máxima r (meses)': [lag_opt_icen, lag_opt_oni],
    'r en ese lag': [max(corrs_icen), max(corrs_oni)],
    'Mediana anomalía El Niño (°C)': [datos_icen[2].median(), datos_oni[2].median()],
    'Mediana anomalía La Niña (°C)': [datos_icen[0].median(), datos_oni[0].median()],
    'p Kruskal-Wallis': [resultados_fases['ICEN']['p_kw'], resultados_fases['ONI']['p_kw']],
}, index=['ICEN', 'ONI']).T
print(resumen.round(3))

# Contraste de la hipótesis exploratoria
crit1 = (datos_icen[2].median() > 0 and datos_icen[0].median() < 0
         and datos_oni[2].median() > 0 and datos_oni[0].median() < 0)
crit2 = abs(r_icen) > abs(r_oni)
print("\nCriterio (i) mediana positiva en El Niño y negativa en La Niña (ICEN y ONI):", "SE CUMPLE" if crit1 else "NO SE CUMPLE")
print("Criterio (ii) |r| con ICEN mayor que |r| con ONI:", "SE CUMPLE" if crit2 else "NO SE CUMPLE")
```

## **Discusión y conclusiones**

1. En el dominio de Piura, la temperatura media a 2 m va de 20,99 °C en febrero a 18,47 °C en julio. En el mapa promedio supera 21 °C en la llanura costera y baja de 16 °C en las zonas altas del este. Las anomalías mensuales tienen una desviación estándar de 0,50 °C y las medias anuales de 0,39 °C. El año más cálido fue 2023 (+0,91 °C) y el más frío, 2000 (−0,68 °C), dos años con fases opuestas de ENSO (Resultado 1). La tendencia de +0,32 °C por década indica que parte de la variabilidad corresponde a un calentamiento de fondo.

2. La anomalía se relaciona de forma positiva y con fuerza parecida con ambos índices. Sin desfase, la correlación de Pearson es de 0,693 con el ICEN y de 0,685 con el ONI (p < 0,001), es decir, cada índice comparte casi la mitad de la varianza de la anomalía (Resultado 3). La relación se mantiene al quitar la tendencia (0,709 y 0,724) y es máxima con un mes de desfase (0,736 y 0,697). Esto indica que la temperatura del aire responde alrededor de un mes después del océano (Resultados 4 y 7).

3. Por fases, la mediana de la anomalía es positiva en El Niño (+0,34 °C con el ICEN y +0,38 °C con el ONI) y negativa en La Niña (−0,37 y −0,34 °C). Las diferencias entre fases son significativas con ambos índices (Kruskal-Wallis, p < 0,001). Según Mann-Whitney, un mes de El Niño fue más cálido que uno de La Niña en el 92% de los pares posibles con el ICEN y en el 91% con el ONI (Resultado 5). La respuesta es asimétrica. Con el ICEN, la desviación estándar es de 0,53 °C en El Niño frente a 0,31 °C en La Niña, porque las anomalías superiores a +1 °C se concentran en 2015-2016 y 2023-2024.

4. La hipótesis se cumple de forma parcial. El criterio (i) se cumple con ambos índices. El criterio (ii) se cumple en los términos definidos, aunque por solo 0,008 (0,693 frente a 0,685). La ventaja se invierte con Kendall (0,443 frente a 0,457) y al quitar la tendencia (0,709 frente a 0,724). El ONI también separa mejor las fases (H de Kruskal-Wallis de 102,87 frente a 92,92). En marzo y abril de 2024, por ejemplo, la anomalía superó +1 °C cuando el ICEN ya estaba en rango neutro y el ONI todavía marcaba El Niño. Solo con un mes de desfase el ICEN queda claramente por encima (0,736 frente a 0,697). Los datos confirman el signo esperado de las anomalías en cada fase, pero no permiten afirmar que la temperatura del dominio dependa más del índice costero.


**Limitaciones:**
* El dominio es un rectángulo que incluye a Piura y parte de los territorios vecinos, por lo que mezcla costa, desierto y sierra.
+ ERA5-Land es un reanálisis de 9 a 11 km de resolución y puede no representar bien la temperatura local en la cordillera Occidental.
+ La climatología de referencia (2000-2025) incluye la tendencia. Como la serie empieza con La Niña (2000) y termina después de El Niño de 2023-2024, la pendiente puede estar inflada.
+ Las fases se asignan mes a mes con umbrales, sin el requisito de persistencia de los episodios oficiales.
+ Las series están autocorrelacionadas y en 26 años hay solo entre ocho y once eventos de cada fase, así que los p-valores son orientativos y las conclusiones, exploratorias.
+ La diferencia entre las correlaciones con el ICEN y con el ONI no se sometió a una prueba estadística.
